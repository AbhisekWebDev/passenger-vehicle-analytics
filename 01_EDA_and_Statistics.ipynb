```python
import numpy as np
import pandas as pd 

df = pd.read_excel('used_cars_data.csv.xlsx')

# print(df.head())

# DATA CLEANSING

# Custom function to clean strings like '19.67 kmpl' or '126.2 bhp' into pure floats
def extract_num(val):
    if pd.isna(val) or str(val).strip().lower() == 'null bhp':
        return np.nan
    return float(str(val).split(' ')[0])

# Apply the cleaning function to create numeric columns
df['Mileage_num'] = df['Mileage'].apply(extract_num)
df['Engine_CC'] = df['Engine'].apply(extract_num)
df['Power_bhp'] = df['Power'].apply(extract_num)

# Handle missing values (Nulls) using Median Imputation
for col in ['Mileage_num', 'Engine_CC', 'Power_bhp', 'Seats', 'Price']:
    df[col] = df[col].fillna(df[col].median())

print("Data cleaning complete! Preview of cleaned numeric columns:")
display(df[['Name', 'Year', 'Fuel_Type', 'Mileage_num', 'Engine_CC', 'Power_bhp', 'Price']].head())

df.to_csv('cleaned_used_cars.csv', index=False)
print("Saved cleaned_used_cars.csv!")
```
